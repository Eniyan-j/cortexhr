```python

### Payroll Data Dictionary Review and Cleaning Decisions

- `payroll_item_id`: Unique identifier for an employee-month pay calculation. Excluded from model inputs.
- `employee_id`: Retained for joins and employee-level grouping; excluded from model inputs.
- `month`: Retained for temporal filtering and aggregation; excluded as a raw model input.
- `run_id`: Payroll-run identifier; excluded from model inputs.
- Duplicate employee-month pairs: 15 pairs were identified. All matched across every column except `payroll_item_id`. These records are preserved and flagged as potential payroll anomalies rather than deleted.
- `variable_pay_type`: 6,202 missing values were found. Every missing-type record has zero `variable_pay`. In the derived dataset, missing types are represented as `"None"`.
- Source preservation: The original workbook is unchanged. All transformations are applied to derived DataFrames.
- Temporal safety: Payroll records must be filtered to the relevant observation period before generating attrition features.
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
```

```python
#numerical column
```

```python
numeric_pipeline=Pipeline([("imputer",SimpleImputer(strategy="median")),
                           ("encoder",StandardScaler())
                          ])
```

```python
categorical_pipeline=Pipeline([("imputer",SimpleImputer(strategy="most_frequent")),
                               ("encoder",OneHotEncoder(handle_unknown="ignore"))
                              ])
```

```python
preprocessor=ColumnTransformer(
    transformers=[
        ("num",numeric_pipeline,numeric_columns),
        ("cat",categorical_pipeline,categorical_columns)],remainder="drop")
```

```python
X_train_processed=preprocessor.fit_transform(X_train)
X_test_processed=preprocessor.transform(X_test)
```

```python
import pandas as pd

# Work on copies; preserve the original datasets
attrition_base = df.copy()
attendance_base = df1.copy()
payroll_base = df2_clean.copy()

# Ensure date columns use datetime
attrition_base["observation_date"] = pd.to_datetime(
    attrition_base["observation_date"], errors="coerce"
)
attendance_base["month"] = pd.to_datetime(
    attendance_base["month"], errors="coerce"
)
payroll_base["month"] = pd.to_datetime(
    payroll_base["month"], errors="coerce"
)

# Keep only attendance and payroll fields that can be aggregated.
# Do not use employee IDs, payroll item IDs, or run IDs as features.
attendance_cols = [
    c for c in attendance_base.select_dtypes(include="number").columns
    if c != "employee_id"
]
payroll_cols = [
    c for c in payroll_base.select_dtypes(include="number").columns
    if c != "employee_id"
]

# Join monthly records to each employee's observation date.
# Only months on or before the observation date are eligible.
def summarize_last_3_months(observations, monthly, value_cols, prefix):
    result = observations[["employee_id", "observation_date"]].copy()
    result["_row_id"] = range(len(result))

    monthly = monthly.dropna(
        subset=["employee_id", "month"]
    ).copy()

    # Avoid using duplicate employee-month records more than once.
    # Numeric values are aggregated by month before selecting the latest 3 months.
    monthly = (
        monthly.groupby(["employee_id", "month"], as_index=False)[value_cols]
        .mean()
    )

    merged = result.merge(
        monthly,
        on="employee_id",
        how="left"
    )

    merged = merged[
        merged["month"].isna()
        if "month" in merged.columns else
        merged["observation_date"].isna()
    ] if False else merged

    return merged

print("Numeric attendance columns:", attendance_cols)
print("Numeric payroll columns:", payroll_cols)
```
