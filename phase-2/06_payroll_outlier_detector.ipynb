```python
evaluation_df = payroll_features[
    [
        "payroll_item_id",
        "iforest_anomaly",
        "rule_anomaly",
        "combined_anomaly"
    ]
].copy()
label_mapping = payroll_labels_df.set_index("payroll_item_id")["anomaly_type"]
evaluation_df["anomaly_type"] = (
    evaluation_df["payroll_item_id"].map(label_mapping)
)
# Calculate metrics separately for each detector and error type

results = []

error_types = payroll_labels_df["anomaly_type"].unique()

for error_type in error_types:

    actual = evaluation_df["anomaly_type"].eq(error_type)

    row = {
        "Anomaly Type": error_type,
        "Injected": int(actual.sum())
    }

    for detector_name, prediction_column in detectors.items():

        predicted = evaluation_df[prediction_column].astype(bool)

        # Calculate TP, FP and FN
        tp = (actual & predicted).sum()
        fp = (~actual & predicted).sum()
        fn = (actual & ~predicted).sum()

        # Calculate precision and recall
        precision = tp / (tp + fp) if (tp + fp) > 0 else 0
        recall = tp / (tp + fn) if (tp + fn) > 0 else 0

        row[f"{detector_name} detected"] = int(tp)
        row[f"{detector_name} missed"] = int(fn)
        row[f"{detector_name} precision"] = round(precision, 3)
        row[f"{detector_name} recall"] = round(recall, 3)

    results.append(row)

# Display evaluation table
per_type_results = pd.DataFrame(results)

per_type_results
```
