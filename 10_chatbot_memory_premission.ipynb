```python
mock_data_dir = Path(
    "/workspace/05_Starter/mock_services/mock_data"
)
employees = pd.read_csv(
    mock_data_dir / "employees.csv"
)
employee_names = sorted(
    employees["full_name"].dropna().unique(),
    key=len,
    reverse=True
)
departments = [
    "Sales",
    "Engineering",
    "Support",
]
job_titles = [
    "Data Analyst",
    "Software Engineer"
]
month_names = [
    "January", "February", "March", "April",
    "May", "June", "July", "August",
    "September", "October", "November", "December"
]
def extract_slots(message):
    text = str(message).strip()
    text_lower = text.lower()
    slots = {}
    for month in month_names:

        if re.search(
            rf"\b{month.lower()}\b",
            text_lower
        ):
            slots["month"] = month
            break

    if re.search(r"\blast month\b", text_lower):
        slots["month_reference"] = "last_month"

    elif re.search(r"\bthis month\b", text_lower):
        slots["month_reference"] = "this_month"
    for name in employee_names:
        if re.search(rf"\b{re.escape(name.lower())}\b", text_lower):

            match_count = (
                employees["full_name"]
                .astype("string")
                .str.casefold()
                .eq(name.casefold())
                .sum()
            )
    
            if match_count > 1:
                slots["employee_name_ambiguous"] = name
            else:
                slots["employee_name"] = name
    
            break
    for department in departments:

        if re.search(
            rf"\b{re.escape(department.lower())}\b",
            text_lower
        ):
            slots["department"] = department
            break

    for title in job_titles:

        if re.search(
            rf"\b{re.escape(title.lower())}\b",
            text_lower
        ):
            slots["job_title"] = title
            break
    leave_patterns = {
        r"\bcasual leaves?\b": "casual",
        r"\bsick leaves?\b": "sick",
        r"\bearned leaves?\b": "earned"
    }
    for pattern, leave_type in leave_patterns.items():
        if re.search(pattern, text_lower):
            slots["leave_type"] = leave_type
            break

    return slots
slot_test_messages = [
    "Is my March payslip ready?",
    "Show my attendance for last month",
    "What is Priya Nair salary?",
    "Who is at risk of leaving in Sales?",
    "And in Engineering?",
    "Suggest a salary band for Data Analyst",
    "What about Software Engineer?",
    "For Support only"
]

for message in slot_test_messages:

    print("Message:", message)
    print("Slots:", extract_slots(message))
    print("-" * 50)
```

```python
session_store = {}
```

```python
def resolve_intent(message, predicted_intent, session_id):
    text = str(message).strip().lower()
    state = session_store.get(
        session_id,
        {"last_intent": None, "slots": {}}
    )
    previous_intent = state["last_intent"]
    resolved_intent = predicted_intent
    follow_up_rules = [
        (r"^(and sick leave|and casual leave|and earned leave)\??$",
         "leave_balance"),

        (r"^and in .+\??$",
         "attrition_risk_query"),

        (r"^why is it high\??$",
         "attrition_risk_query"),

        (r"^show details\??$",
         "payroll_flags"),

        (r"^what about .+\??$",
         "salary_band_query"),

        (r"^for .+ only$",
         "headcount_forecast")
    ]
    for pattern, expected_intent in follow_up_rules:

        if (
            re.fullmatch(pattern, text)
            and previous_intent == expected_intent
        ):
            resolved_intent = previous_intent
            break
    if (
        re.fullmatch(r"(?:and )?(?:last|this) month\??", text)
        and previous_intent in [
            "payslip_status",
            "attendance_summary",
            "payroll_flags"
        ]
    ):
        resolved_intent = previous_intent
    new_slots = extract_slots(message)
    merged_slots = state["slots"].copy()
    if "month_reference" in new_slots:
        merged_slots.pop("month", None)
    elif "month" in new_slots:
        merged_slots.pop("month_reference", None)
    merged_slots.update(new_slots)
    session_store[session_id] = {
        "last_intent": resolved_intent,
        "slots": merged_slots
    }
    return {
        "intent": resolved_intent,
        "slots": merged_slots
    }
```

```python
demo_users = pd.read_csv("/workspace/05_Starter/mock_services/mock_data/demo_users.csv")
def get_demo_role(employee_id):
    matching_users = demo_users[
        demo_users["employee_id"].astype(str) == str(employee_id)
    ]
    if matching_users.empty:
        return None
    return matching_users.iloc[0]["role"].upper()
```

```python
def check_intent_permission(intent, user_role):
    if intent == "salary_of_other":
        return False, "I cannot provide another employee's salary information."
    row = chatbot_responses[chatbot_responses["intent"] == intent]
    if row.empty:
        return False, "Unknown intent. Access denied."
    allowed_roles = str(row.iloc[0]["allowed_roles"])
    allowed_roles = [role.strip().upper()for role in allowed_roles.split(",")]
    user_role = str(user_role).strip().upper()
    if "ALL" in allowed_roles or user_role in allowed_roles:
        return True, "Access granted."
    return False, "You do not have permission to access this information."
```

```python
MOCK_HR_URL = "http://127.0.0.1:9000"
def call_mock_hr_service(intent, user_id, slots=None):
    if slots is None:
        slots = {}
    role = get_demo_role(user_id)

    if role is None:
        return "Authentication failed. Invalid demo user."
    allowed, message = check_intent_permission(intent, role)
    if not allowed:
        return message
    if intent == "salary_of_other":
        return "I cannot provide another employee's salary information."
    headers = {
        "X-Demo-User": user_id
    }
    endpoints = {
        "leave_balance": "/api/v1/leave/balance",
        "leave_request_status": "/api/v1/leave/requests",
        "my_manager": "/api/v1/employees/me",
        "my_department": "/api/v1/employees/me",
        "team_members": "/api/v1/employees/me/team",
        "holiday_list": "/api/v1/holidays"
    }

    if intent in ["greet", "goodbye", "thanks", "chitchat", "help",
                  "out_of_scope"]:
        return "No HR service call required."

    if str(chatbot_responses.loc[
        chatbot_responses["intent"] == intent,
        "backend_or_service_call"
    ].iloc[0]).startswith("faq:"):
        return "FAQ response does not require a mock HR API call."

    if intent not in endpoints and intent not in [
        "payslip_status", "attendance_summary"
    ]:
        return "This service is not available in the mock HR backend."
    try:
        if intent == "leave_balance":
            response = requests.get(
                MOCK_HR_URL + endpoints[intent],
                headers=headers,
                timeout=10
            )
        elif intent == "leave_request_status":
            response = requests.get(
                MOCK_HR_URL + endpoints[intent],
                headers=headers,
                params={"limit": 1},
                timeout=10
            )
        elif intent in ["my_manager", "my_department"]:
            response = requests.get(
                MOCK_HR_URL + endpoints[intent],
                headers=headers,
                timeout=10
            )

        elif intent == "team_members":
            response = requests.get(
                MOCK_HR_URL + endpoints[intent],
                headers=headers,
                timeout=10
            )

        elif intent == "holiday_list":
            response = requests.get(
                MOCK_HR_URL + endpoints[intent],
                headers=headers,
                timeout=10
            )

        elif intent == "payslip_status":
            month = resolve_month_key(slots)

            if not month:
                return "Which month would you like to check? Please provide YYYY-MM."

            response = requests.get(
                f"{MOCK_HR_URL}/api/v1/payslips/me/{month}",
                headers=headers,
                timeout=10
            )

        elif intent == "attendance_summary":
            month = resolve_month_key(slots)

            if not month:
                return "Which month would you like to check? Please provide YYYY-MM."

            response = requests.get(
                MOCK_HR_URL + "/api/v1/attendance/summary",
                headers=headers,
                params={"month": month},
                timeout=10
            )
        if response.status_code == 404:
            return "No matching HR record was found for the requested period."
        response.raise_for_status()
        return response.json()
    except requests.RequestException as error:
        return f"HR service request failed: {error}"
```

```python
from datetime import date, timedelta

def resolve_month_key(slots, today=None):

    if today is None:
        today = date.today()

    if slots.get("month"):
        month_number = {
            "January": 1,
            "February": 2,
            "March": 3,
            "April": 4,
            "May": 5,
            "June": 6,
            "July": 7,
            "August": 8,
            "September": 9,
            "October": 10,
            "November": 11,
            "December": 12
        }
        month = month_number[slots["month"]]
        return f"{today.year}-{month:02d}"
    if slots.get("month_reference") == "last_month":
        previous_month = today.replace(day=1) - timedelta(days=1)
        return previous_month.strftime("%Y-%m")

    # This month
    if slots.get("month_reference") == "this_month":
        return today.strftime("%Y-%m")
    return None
```
